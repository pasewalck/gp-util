```python
import numpy as np
import matplotlib.pyplot as plt
import locale
from jona_gp_util.rounding import din_1333_round,pdg_round
from jona_gp_util.use_figure import use_figure

def din_1333_round_err(error):
    value,error, _,_ = din_1333_round(0,error)
    return error

def pdg_round_err(error):
    value,error, _,_ = pdg_round(0,error)
    return error

locale.setlocale(locale.LC_NUMERIC, 'de_DE.UTF-8')

errors = np.arange(0,1000,0.2)
errors_pdg = np.array([pdg_round_err(x) for x in errors])
errors_din_1333 = np.array([din_1333_round_err(x) for x in errors])

with use_figure() as fig:

    plt.plot(errors,errors_din_1333,linestyle="--",label="DIN-1333")
    plt.plot(errors,errors_pdg,linestyle="--",label="PDG")

    plt.tight_layout()

    plt.legend()
    plt.grid(True)
```
