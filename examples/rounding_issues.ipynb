```python
def display_rounding_results(fn,org_value,org_error,exact=False):
    value,error,decimals,digits = fn(org_value,org_error,exact)
    print(f"Rounded {org_value} ± {org_error}")
    print(f"     to {value:.{max(0, decimals)}f} ± {error:.{max(0, decimals)}f}")
    print("")
```

Ein paar einfache Beispiele wie DIN-1333 rundet.

```python
from jona_gp_util.rounding import din_1333_round

display_rounding_results(din_1333_round,1213, 112)
display_rounding_results(din_1333_round,1213, 212)
display_rounding_results(din_1333_round,1213, 312)
display_rounding_results(din_1333_round,1213, 412)

display_rounding_results(din_1333_round,9181, 12)
display_rounding_results(din_1333_round,9181, 412)
```

Die din_1333_round Funktion hat zwei Modi. Der Modus wird durch den dritten Parameter eingestellt. Siehe `din_1333_round(value,error,exact)`, also `exact`. Dieses Parameter kann wahr oder falsch sein und beeinflusst, ob die Funktion mit floats oder Decimals als Datentyp verwendet bei der Berechnung. Floats haben den Nachteil von Ungenauigkeiten (sog. floating point inaccuracies) sowie, dass die Rundungsfunktion `round(number)` in Python half-to-even ist, sie rundet also $4{,}5$ auf $4$ statt auf $5$. Siehe unten.

```python
display_rounding_results(din_1333_round,4.5, 0.3)
display_rounding_results(din_1333_round,4.5, 0.3,True)
```
